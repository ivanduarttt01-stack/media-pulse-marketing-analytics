# 📣 Media Pulse — Performance Analytics

```
import pandas as pd
df=pd.read_csv('../data/raw/campaign_performance.csv')
df.head()
```

```
ch=df.groupby('channel').agg(spend=('spend_usd','sum'),revenue=('revenue_usd','sum'),conversions=('conversions','sum'))
ch['roas']=ch.revenue/ch.spend
ch['cpa']=ch.spend/ch.conversions
ch.sort_values('roas',ascending=False)
```

## Interpretación

Separar eficiencia de escala antes de proponer cambios de presupuesto.
